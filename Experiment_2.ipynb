```python
%%bash
#!/bin/bash
# In Colab's '%%bash' cells, 'read' does not work interactively.
# Assigning values directly for demonstration purposes.
a=10
b=20
c=5

echo "Comparing numbers: $a, $b, $c"
if [ $a -gt $b ] && [ $a -gt $c ]
then
  echo "$a is greater"
elif [ $b -gt $c ]
then
  echo "$b is greater"
else
  echo "$c is greater"
fi
```
