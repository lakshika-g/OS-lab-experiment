```python
%%bash
#!/bin/bash
echo "ENTER THE LIMIT:"
read n

# Validate if n is an integer
if ! [[ "$n" =~ ^-?[0-9]+$ ]]; then
  echo "Error: Input for LIMIT must be an integer." >&2
  exit 1
fi

p=-1
q=1
i=1
while [ $i -le "$n" ]
do
r=$((p + q))
p=$q
q=$r
echo "$r"
i=$((i + 1))
done
```
