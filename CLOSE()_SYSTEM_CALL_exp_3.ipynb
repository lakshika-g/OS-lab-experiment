To compile and run C code in a Colab notebook, you can use the `%%bash` magic command. This allows you to execute shell commands directly in the cell, including `gcc` for compiling C code.

```python
%%bash
# Write the C code to a file
cat << EOF > sample.c
#include <stdio.h>
#include <fcntl.h>
#include <unistd.h>

int main()
{
    int fd;
    fd = open("sample.txt", O_RDONLY);
    if(fd < 0)
    {
        printf("File Opening Failed\n");
        return 1;
    }
    printf("File Opened Successfully\n");
    close(fd);
    printf("File Closed Successfully\n");
    return 0;
}
EOF

# Compile the C code
gcc sample.c -o sample_program

# Create an empty sample.txt file for the program to open
touch sample.txt

# Execute the compiled program
./sample_program
```

This will first write your C code into a file named `sample.c`, then compile it using `gcc` into an executable named `sample_program`, and finally run the executable. I've also added a `touch sample.txt` command to create the file that your C program tries to open, preventing a file opening failure.
