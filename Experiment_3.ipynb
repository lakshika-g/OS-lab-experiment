```python
#include <stdio.h>
#include <stdlib.h>
#include <unistd.h>
#include <sys/wait.h>
int main()
{
pid_t pid;
pid = fork();
if(pid < 0)
{
printf("Fork Failed\n");
exit(1);
}
else if(pid == 0)
{
printf("\nCHILD PROCESS");
printf("\nChild PID : %d", getpid());
printf("\nParent PID : %d\n", getppid());
exit(0);
}
else
{
wait(NULL);
printf("\nPARENT PROCESS");
printf("\nParent PID : %d", getpid());
printf("\nParent's Parent PID : %d\n", getppid());
}
return 0;
}
```

First, let's save the C code into a file named `fork_example.c`.

```python
c_code = r'''
#include <stdio.h>
#include <stdlib.h>
#include <unistd.h>
#include <sys/wait.h>

int main()
{
pid_t pid;
pid = fork();
if(pid < 0)
{
printf("Fork Failed\n");
exit(1);
}
else if(pid == 0)
{
printf("\nCHILD PROCESS");
printf("\nChild PID : %d", getpid());
printf("\nParent PID : %d\n", getppid());
exit(0);
}
else
{
wait(NULL);
printf("\nPARENT PROCESS");
printf("\nParent PID : %d", getpid());
printf("\nParent's Parent PID : %d\n", getppid());
}
return 0;
}
'''

with open('fork_example.c', 'w') as f:
    f.write(c_code)

print("C code saved to fork_example.c")
```

Next, we'll compile the C code using `gcc`. We'll use the `%%bash` magic command to run shell commands.

```python
%%bash
gcc fork_example.c -o fork_example

if [ $? -eq 0 ]; then
  echo "Compilation successful!"
else
  echo "Compilation failed."
  exit 1
fi
```

Finally, let's run the compiled program.

```python
%%bash
./fork_example
```
