```python
%%bash
# Assign a test string directly since 'read' doesn't work interactively in Colab cells
str="madam"
# Or to test with a non-palindrome: str="hello"

echo "Checking if '$str' is a palindrome..."

len=$(echo -n "$str" | wc -c)
i=1
j=$((len / 2))

is_palindrome=true

while [ $i -le $j ]
do
k=$(echo "$str" | cut -c $i)
l=$(echo "$str" | cut -c $((len - i + 1)))

if [ "$k" != "$l" ]
then
is_palindrome=false
break
fi
i=$((i + 1))
done

if [ "$is_palindrome" = true ]
then
echo "'$str' is a palindrome"
else
echo "'$str' is not a palindrome"
fi
```
