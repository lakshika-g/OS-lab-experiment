```python
%%bash
#!/bin/bash
echo "ENTER THE RANGE:"
n=10 # Assigning a default value for n, as 'read n' doesn't work interactively here
x=1
sum=0
while [ $x -le $n ]
do
sum=$((sum + x))
x=$((x + 2))
done

echo "SUM = $sum"
```
